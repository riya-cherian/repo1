```python
import pandas as pd

data = {
    'customer': ['A', 'B', 'C', 'A', 'D', 'B', 'E', 'C', 'A', 'D'],
    'order_date': [
        '2026-01-05', '2026-01-10', '2026-01-12', '2026-01-15',
        '2026-02-02', '2026-02-10', '2026-02-15', '2026-02-20',
        '2026-03-03', '2026-03-10'
    ],
    'region': ['East', 'West', 'East', 'East', 'West', 'West', 'East', 'East', 'East', 'West'],
    'product': ['Laptop', 'Phone', 'Laptop', 'Headphones', 'Phone', 'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'sales': [1200, 800, 1500, 200, 900, 1100, 600, 750, 1300, 700],
    'quantity': [1, 1, 1, 2, 1, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

df.head()
```

```python
df.groupby('product').agg(
    TotalOrders = ('product','count')
)
```

```python
df.groupby('product').agg(
    TotalSales = ('sales','sum')
)
```

```python
customersales = df.groupby('customer').agg(
    total_sales = ('sales','sum')
)
customersales.sort_values(by=['total_sales'], ascending =False).head(1)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df
```

```python
df['month']= df['order_date'].dt.month
df
```

```python
 df.groupby('month').agg(
    total_sales = ('sales','sum')
).nlargest(1,'total_sales')
```

```python
df.groupby('month').agg(
    total_sales = ('sales','sum')
).sort_values('total_sales', ascending=False).head(1)
```

```python
tcs = df.agg({'sales':['sum']})
tcs
```

```python
product_sales = df.groupby('product').agg(
    total_pdt_sales = ('sales','sum')
)
product_sales
```

```python
product_sales['sales_percentage'] = (product_sales['total_pdt_sales']*100)/tcs['sales']['sum']
product_sales
```
